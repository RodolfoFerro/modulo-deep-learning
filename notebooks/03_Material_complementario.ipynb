```python
import tensorflow as tf
import numpy as np


# Sample data (XOR)
x = np.array([(0, 0), (1, 0), (0, 1), (1, 1)])
y = np.array([0, 1, 1, 0])


# Creates the model
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(2, activation='linear'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])


# Compile the model
loss = tf.keras.losses.MeanSquaredError()
optimizer = tf.keras.optimizers.SGD(learning_rate=0.6)

model.compile(optimizer=optimizer, loss=loss, metrics=[loss])
```

Creamos el callback especificando el valor épsilon para la diferencia de valores (es decir, especificamos los decimales a considerar en el criterio de paro).

```python
# Create a callback
callback = tf.keras.callbacks.EarlyStopping(
    monitor='loss',
    min_delta=1e-4, # Este valor define el ε (la diferencia)
    patience=3
)
```

```python
history = model.fit(x, y, epochs=1000, callbacks=[callback])
```

```python
import plotly.express as px


losses = history.history['loss']
eje_x = np.arange(len(losses))

fig = px.line(
    x=eje_x,
    y=losses,
    title='Historia de entrenamiento',
    labels=dict(x='Épocas', y='Error')
)
fig.show()
```

## **Sección II – Ejercicio**

### **Clasificación 01**

Para este problema utilizaremos un dataset sintético ya conocido:

```python
from sklearn.datasets import make_circles
import numpy as np


# Make 1000 examples
n_samples = 1000

# Create circles
x, y = make_circles(n_samples, noise=0.03, random_state=42)
```

```python
import pandas as pd


circles = pd.DataFrame({"x1": x[:, 0], "x2": x[:, 1], "label":y})
circles.head()
```

```python
# Verify number of labels
circles.label.value_counts()
```

```python
import plotly.express as px


fig = px.scatter(
    x=circles['x1'],
    y=circles['x2'],
    color=circles['label'].astype(str) # Discretize the color palette
)
fig.show()
```

Transformamos los datos utilizando el cuadrado de las características.

```python
x_sq = x ** 2
```

```python
circles_sq = circles.copy()
circles_sq["x1_sq"] = circles_sq["x1"] ** 2
circles_sq["x2_sq"] = circles_sq["x2"] ** 2
```

```python
import plotly.express as px

fig = px.scatter(
    x=circles_sq['x1_sq'],
    y=circles_sq['x2_sq'],
    color=circles_sq['label'].astype(str)
)
fig.show()
```

Como podemos ver, esta transformación nos permite tener un conjunto cuyas características son linealmente separables, simplificando el problema y pudiendo utilizar una simple neurona para resolver el problema.

### **Optimizador y función de pérdida**

$$ \mathrm{Binary Crossentropy}=-\frac{1}{N}\cdot\sum_{i=1}^N \left[y_i \log (\hat{y}_i) + (1 - y_i) \log (1 - \hat{y}_i) \right] $$


```python
# TODO: Create loss function and optimizer
loss = tf.keras.losses.BinaryCrossentropy()
optimizer = tf.keras.optimizers.Adam(learning_rate=3e-3) # Try Adam(learning_rate=3e-3)
```

```python
# TODO: Integrate all here!
model.compile(loss=loss, optimizer=optimizer, metrics=['accuracy'])
```

### **Evaluación del modelo**

```python
model.evaluate(x_sq, y)
```

```python
y_pred = model.predict(x_sq)
```

```python
def binarize_output(y_pred, threshold=0.5):
    return np.array([1 if x > threshold else 0 for x in y_pred])
```

```python
from sklearn.metrics import accuracy_score


y_pred_bin = binarize_output(y_pred)
print("Accuracy score:", accuracy_score(y, y_pred_bin))
```
